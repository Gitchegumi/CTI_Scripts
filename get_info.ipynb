```python
import requests
import os
from dotenv import load_dotenv
```

```python
def get_api_token_and_uuid(email, password, broker_id, token):
    """Fetch API token and account UUID by logging in."""
    url = "https://platform.citytradersimperium.com/mtr-api/manager/co-login"
    payload = {
        "email": email,
        "password": password,
        "brokerId": broker_id
    }
    headers = {
        "Authorization": f"Bearer {token}",
        "Accept": "application/json", 
        "Content-Type": "application/json" 
    }
    session = requests.Session()
    session.cookies.set("cf_clearance", "value_from_browser", domain="platform.citytradersimperium.com")
    response = session.post(url, json=payload, headers=headers)
    if response.status_code == 200:
        data = response.json()
        api_token = data.get("token")
        accounts = data.get("accounts", [])
        if accounts:
            account_uuid = accounts[0].get("tradingAccountId")
            return api_token, account_uuid
        else:
            print("No accounts found in response.")
            return None, None
    else:
        print(f"Login failed: {response.status_code} {response.text}")
        return None, None
```

```python
load_dotenv()

EMAIL = os.getenv("EMAIL")
PASSWORD = os.getenv("PASSWORD")
BROKER_ID = os.getenv("BROKER_ID")
TOKEN = os.getenv("TOKEN")

# Fetch API token and UUID
API_TOKEN, ACCOUNT_UUID = get_api_token_and_uuid(EMAIL, PASSWORD, BROKER_ID, TOKEN)

if API_TOKEN and ACCOUNT_UUID:
    print(f"API Token: {API_TOKEN}")
    print(f"Account UUID: {ACCOUNT_UUID}")
else:
    print("Failed to retrieve API token and UUID.")
```
