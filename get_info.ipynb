```python
def refresh_token():
    """Refresh the authentication token."""
    auth_trading_api = login_manager.auth_trading_api
    url = f"{utils.PLATFORM_URL}/manager/refresh-token?rt={auth_trading_api}"
    headers = {
        "Accept": "application/json",
        "Content-Type": "application/json",
        "User-Agent": "Mozilla/5.0",
    }
    try:
        response = requests.post(url, headers=headers, timeout=10)
        response.raise_for_status()
        print(response)
        try:
            new_token = response.json().get("token")
            if new_token:
                auth_trading_api = new_token
                print("Token refreshed successfully.")
            else:
                print("Failed to refresh token: No token in response.")
        except ValueError:
            print("Failed to refresh token: Response is not in JSON format.")
    except requests.exceptions.RequestException as e:
        print("Failed to refresh token: %s", e)
```

```python
refresh_token()
```
